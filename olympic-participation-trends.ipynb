```python
import pandas as pd

# Athlete-level Olympic records, 1896–2016 (Kaggle: "120 years of Olympic history")
df = pd.read_csv("athlete_events.csv")
df.head()
```

```python
# Introduction

# This dataset contains Olympic athlete data from 1896 to 2016, covering over 120 years of Olympic history.

# The dataset includes information about athletes, their countries, the events they participated in, and the medals they won. 
# By analyzing this data, we can uncover patterns related to gender participation, country performance, and medal distribution across sports.

# The goal of this analysis is to answer the following questions:
# How has female participation in the Olympics changed over time?
# Which countries have won the most Olympic medals?
# Which sports award the most medals?

# We will explore these questions and summarize key findings at the end.
```

```python
# Q1. # How has female participation in the Olympics changed over time?

import matplotlib.pyplot as plt

# Count of unique male and female athletes per year
gender_participation = df.drop_duplicates(subset=['Year', 'Name', 'Sex']) \
                         .groupby(['Year', 'Sex'])['Name'].count().unstack()

# Plot
gender_participation.plot(kind='line', figsize=(10,6))
plt.title('Olympic Participation by Gender Over Time')
plt.xlabel('Year')
plt.ylabel('Number of Athletes')
plt.grid(True)
plt.legend(title='Sex')
plt.tight_layout()
plt.show()
```

```python
# This graph shows the number of male and female athletes participating in each Olympic year. 
# We can see that while male participation has always been higher, female participation has grown, especially after 1980. 
# By 2016, the gender gap had narrowed.
```

```python
# Q2. # Which countries have won the most Olympic medals?

# Filter out rows with medals
medals = df[df['Medal'].notnull()]

# Count medals by team
top_countries = medals['Team'].value_counts().head(10)

# Plot
top_countries.plot(kind='bar', figsize=(10,6))
plt.title('Top 10 Countries by Total Olympic Medals')
plt.xlabel('Country')
plt.ylabel('Number of Medals')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# This bar chart shows the top 10 countries based on total medals won. 
# The United States leads by a large margin, followed by countries like the Soviet Union, Germany, and Great Britain. 
# This reflects both athletic performance and participation levels.
```

```python
# Q3. # Which sports award the most medals?

# Count medals by sport
top_sports = medals['Sport'].value_counts().head(10)

# Plot
top_sports.plot(kind='barh', figsize=(10,6))
plt.title('Top 10 Sports by Total Olympic Medals Awarded')
plt.xlabel('Number of Medals')
plt.ylabel('Sport')
plt.tight_layout()
plt.show()
```

```python
# This horizontal bar chart shows the sports that have awarded the most medals. 
# Athletics and swimming dominate due to the large number of events they have. 
# These sports offer many opportunities for athletes to medal, unlike others with fewer events.
```

```python
# Conclusion

# This analysis of Olympic history from 1896 to 2016 revealed several insights:

# Female participation in the Olympics has grown steadily over time, closing the gap with male athletes in recent years.
# The United States has won the most Olympic medals by far, with a consistent presence across multiple decades.
# Sports like athletics and swimming dominate the medal counts due to the number of events they include.

# The Olympics provide a unique window into global athletic performance, national investment in sports, and progress in gender equality over time.
```
